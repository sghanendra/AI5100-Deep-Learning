```python
#Understanding Error Surface
#All imports
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import collections

########################################
#Generate meaningfull data
########################################
np.random.seed(102)

heights = np.linspace(30,80,25,endpoint=True)
noise = np.random.normal(0,20,len(heights))
weights = 3.86*heights - 110.42 + np.abs(noise)
########################################
#Plot scatter plot of data
########################################
plt.scatter(heights,weights,label="Data")
plt.xlabel("Heights (Inches)")
plt.ylabel("Weights (Pounds)")
#plt.show()

########################################
#Weight estimation through error surface, i.e., empirically locate the minima of error surface
########################################

#Compute the error
def Error(w,t,x): #inputs : 1)weight 2)data i.e (t,x)
    #Estimate the target
    target = w[0]*x + w[1]
    error = np.square(t - target)
    #Compute and return the error
    return error


#Plot 3D error surface and the corresponding contour plots
#Error surface is a function of w0 and w1

########################################
#Least squares approach to estimate the weights
########################################
#Complete the below linear regression function
def LinearRegression(x,t): #inputs : 1)input data i.e (x). 2)target i.e (t)
    X = np.vstack([x, np.ones(len(x))]).T
    t = t[:, np.newaxis] #y a column vector
    w_opt = np.dot((np.dot(np.linalg.inv(np.dot(X.T,X)),X.T)),t)
    return w_opt

#Estimate optimal weight's using "LinearRegression" function
weights_opt = LinearRegression(heights, weights)
#print("Optimal Weights: ", weights_opt)

#Estimate the targets using the input x and the estimated weights
targets = weights_opt[0]*heights + weights_opt[1]

#Plot the estimated line on top of the above scatter plot
plt.plot(heights, targets, 'r',label="Estimate")
plt.legend()
plt.show()

########################################
#Compare the estimated weight's using least squares approach with the error surface approach
########################################

w0,w1 = np.meshgrid(heights,weights)
E = Error(weights_opt,w1,w0)
plt.contourf(w0,w1,E)
plt.show()

ax = plt.axes(projection='3d')
ax.scatter(w0,w1,E)
ax.set_xlabel("Heights (Inches)")
ax.set_ylabel("Weights (Pounds)")
ax.set_zlabel("Error")
plt.tight_layout()
plt.show()
```

```python
#Understanding model order and overfitting
########################################
#Generate 20 data points
########################################

D = 20
print(f"Data Size: {D}")
def gen_data(D):
    np.random.seed(3)
    X = np.linspace(0,1,D)
    e = np.random.normal(0,0.1,D)
    Y = np.sin(2*np.pi*1*X) + e
    return X,Y

X,Y = gen_data(D)

#Obtain train and test splits
X_train = X[0:D:2]
Y_train = Y[0:D:2]
X_test = X[1:D:2]
Y_test = Y[1:D:2]


########################################
#Fitting Mth degree polynomial using least squares approach
########################################
#Complete the function
def PolynomialFit(X_train,Y_train,M,lamda): #(training data, training targets, Model order, Regularization coefficient)
    #Transform the data using polynomial kernel
    A = np.linalg.inv(np.array([[sum(np.power(X_train, i)) for i in range(j, M + j)] for j in range(M)]) + lamda*np.identity(M)) 
    #Find Pseudo inverse solution
    w_opt = np.dot(A, np.array( [ sum(Y_train * np.power(X_train, i)) for i in range(M)])) 
    #return the weight vector
    return w_opt

#Complete the function
def PolynomialPred(w_opt,X_train,X_test): #(weight,training data, testing data, training targets, testing targets)
    #Estimate the targets for both training and testing data    

    TrainPred = sum([w_opt[i] * X_train**i for i in range(len(w_opt))])
    TestPred = sum([w_opt[i] * X_test**i for i in range(len(w_opt))])
    
    #Return training and testing predictions
    return TrainPred,TestPred

#Complete the function
def PolynomialPred_Error(w_est,X_train,X_test,Y_train,Y_test): #(weight,training data, testing data, training targets, testing targets)
    #Estimate the targets for both training and testing data

    Train_pred, Test_pred = PolynomialPred(w_est, X_train,X_test)
    
    TrainError = np.sqrt(np.mean((Y_train-Train_pred)**2))
    TestError = np.sqrt(np.mean((Y_test-Test_pred)**2))
    
    #Return training and testing error
    return TrainError,TestError

#Iterate through range of M values
lamda = 0
M_range = 10
TrError = []
TeError = []
TrPred = []
TePred = []

for M in range(1,M_range):
    
    #Fit Mth order polynomial i.e estimate optimal w. Use the function "PolynomialFit"
    w_est = PolynomialFit(X,Y,M,lamda)
    #Predict training and testing targets
    Tr_pred, Te_pred = PolynomialPred(w_est, X_train, X_test)
    TrPred.append(Tr_pred)
    TePred.append(Te_pred)
    
    #Predict errors on both training and testing data using estimated w. Use the function "PolynomialPred_Error"
    Tr_err, Te_err = PolynomialPred_Error(w_est,X_train,X_test,Y_train,Y_test)
    #Store them for plotting
    TrError.append(Tr_err)
    TeError.append(Te_err)
    
w_est = PolynomialFit(X,Y,9,lamda)
Tr_p,Te_p = PolynomialPred(w_est,X,X)
plt.scatter(X, Y,label='Data')
plt.plot(X, Tr_p, c="red",label='Fit M=%d'%M)
plt.legend()
plt.show()

#Plot training and testing estimates alogwith the original targets
for i in range(len(TrPred)):
    plt.subplot(2,2,1)
    plt.title("Training")
    plt.plot(X_train,TrPred[i],label='Tr_Pred')
    plt.plot(X, Y,'k.',label='Data')
    plt.subplot(2,2,2)    
    plt.title("Testing")
    plt.plot(X_test,TePred[i],label='Te_Pred')
    plt.plot(X, Y,'k.',label='Data')

#plt.xlabel(r"$X$")
#plt.ylabel(r"$Y$")

#Plot training error vs polynomial order, and testing error vs polynomial order
plt.suptitle("Estimate and Error Plots")
plt.subplot(2,2,3)
plt.plot(TrError,label='Tr_Err')
plt.plot(TrError,'k.')
#plt.plot(X, Y,'k.',label='Data')
plt.xlabel("Polynomial Order (M)")
plt.ylabel("Error")
plt.subplot(2,2,4)    
plt.plot(TeError,label='Te_Err')
plt.plot(TeError,'k.')
#plt.plot(X, Y,'k.',label='Data')
plt.xlabel("Polynomial Order (M)")
plt.ylabel("Error")
plt.show()
```

```python
#Understanding the bias-variance trade-off
########################################
#Generate 100 data sets of noisy sinusoidal data
########################################

np.random.seed(222)
d_size = 100
fn = np.linspace(0,1,d_size)
sn = np.sin(2*np.pi*fn) + np.random.normal(0,0.1,d_size)

def gaussian(xn,sig):
    return np.exp(-0.5*(xn-np.mean(xn)/sig)**2)

########################################
#Use regularized least squares to estimate w
########################################
def reg_Least_Square(fn,sn,reg):  

    fn = gaussian(fn,0.72)    
    Xf = np.linalg.inv(np.array([[sum(np.power(fn, i)) for i in range(j, d + j) ] for j in range(d)]) +reg*np.identity(d))
    Wf = np.dot(Xf, np.array( [ sum(sn * np.power(fn, i)) for i in range(d)]))    
    Tf = sum([Wf[i] * fn**i for i in range(len(Wf))])
    return Tf

d = 25
reg = 0.0
T_ls = reg_Least_Square(fn,sn,0.0)
plt.plot(fn,T_ls,'r',label=r"Least Square LS ($\lambda$=%.1f)"%reg)
reg = 1e-2
T_rls = reg_Least_Square(fn,sn,reg)
plt.plot(fn,T_rls,'g', label=r"Regularized LS RLS ($\lambda$=%.2f)"%reg)
plt.plot(fn,sn,'k.',label="Data")
plt.legend()
plt.show()
```

```python
########################################
#Illustrate the concept of Bias-Variance trade off
########################################
#1. Chose three different regularization coefficeints (low,middle and high)

#2. For every regularization coefficient, produce two plots: one displaying 100 estimated curves, and the other showing the mean of the estimated curves alongside the original function.
l1 = np.linspace(1e-4,1e-3,100)
l2 = np.linspace(1e-0,1e1,100)
l3 = np.linspace(1e-2,1e3,100)

#3. For three regularization coefficients, you should have a total of six plots, meaning two plots for each regularization.
#4. Using the six plots above, describe the bias-variance trade-off.

L1 = []
plt.figure(figsize=(12,6))
plt.suptitle("Regularized Regression")
plt.subplot(2,3,1)
for reg_c in l1:
    T_reg = reg_Least_Square(fn,sn,reg_c)
    L1.append(T_reg)
    plt.plot(fn,T_reg)
plt.legend([r"($\lambda$=low)"])

plt.subplot(2,3,4)
plt.plot(fn,sn,'k.',label="Data")
plt.plot(fn,np.mean(L1,axis=0),'r',label=r'$\mu$')
plt.plot(fn,T_ls,'g', label=r"($\lambda$)")
plt.legend()
#plt.show()

L2 = []
plt.subplot(2,3,2)
for reg_c in l2:
    T_reg = reg_Least_Square(fn,sn,reg_c)
    L2.append(T_reg)
    plt.plot(fn,T_reg)
plt.legend([r"($\lambda$=mid)"])
plt.subplot(2,3,5)
plt.plot(fn,sn,'k.',label="Data")
plt.plot(fn,T_ls,'g', label=r"($\lambda$)")
plt.plot(fn,np.mean(L2,axis=0),'r',label=r'$\mu$')
plt.legend()
#plt.show()

L3 = []
plt.subplot(2,3,3)
for reg_c in l3:
    T_reg = reg_Least_Square(fn,sn,reg_c)
    L3.append(T_reg)
    plt.plot(fn,T_reg)

plt.legend([r"($\lambda$=high)"])
plt.subplot(2,3,6)
plt.plot(fn,sn,'k.',label="Data")
plt.plot(fn,T_ls,'g', label=r"($\lambda$)")
plt.plot(fn,np.mean(L3,axis=0),'r',label=r'$\mu$')
plt.legend()
plt.show()
```

<b> Report your observations </b>

1. For no regularization, we might require high degree for less data to fit which results in overfitting the data.
2. Low regularization value supresses weights just enough to fit the data.
3. Very high value of regularization leads to weight suppression resulting in poor fit.



```python
#Understanding MAP estimate
########################################
#Generate 100 data sets of noisy sinusoidal data
########################################
np.random.seed(123)
Xn = np.linspace(0,1,100)
Yn = np.sin(2*np.pi*Xn) + np.random.normal(0,0.05,100)
########################################
#Update the statistics of posterior density
########################################
#Initialie the parameters for standard normal prior
d = 8
sigma = 0.0
alpha = 1.0

def map_estimate(X, y, sigma, alpha,d):

    Phi = np.linalg.inv(np.array([[sum(np.power(X, i)) for i in range(j, d + j) ] for j in range(d)]) +(sigma/alpha)**2*np.identity(d))    
    theta = np.dot(Phi, np.array([ sum(y * np.power(X, i)) for i in range(d)]))
    YPred = sum([theta[i] * X**i for i in range(len(theta))])    

    return theta, YPred


#Iterate through the data points and update the stats of posterior density
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.title("MAP")
Theta, Ypred = map_estimate(Xn,Yn,sigma,alpha,d)
plt.plot(Xn,Ypred,label=r'MAP $\sigma$=%0.1f & $\alpha$=%0.1f'%(sigma,alpha))
plt.plot(Xn,Yn,'k.')
sigma = 0.5
Theta, Ypred = map_estimate(Xn,Yn,sigma,alpha,d)
plt.plot(Xn,Ypred,label=r'MAP $\sigma$=%0.1f & $\alpha$=%0.1f'%(sigma,alpha))
plt.plot(Xn,Yn,'k.')
plt.legend(loc='upper right')
plt.xlabel("X")
plt.ylabel("Y")
#plt.show()

########################################
#Sample weight vector from posterior distribution. Estimate the curve, repeat the procedure for 100 times and get the avg fit
########################################

########################################
#Predictive distribution analysis
########################################
#Predictive distribution analysis through sampling
#Iterate through data points and sample weight vectors when partial data points are seen, and plot the curves

sigma = 0.0
plt.subplot(1,3,2)
Y_avg = []
for i in range(100):

    TH = np.random.choice(Theta, 100, replace=True)
    Yp = sum([TH[i] * Xn**i for i in range(len(TH))])
    TH,YP = map_estimate(Xn,Yp,sigma,alpha,d)
    Y_avg.append(YP)
    plt.plot(Xn,YP/10)

plt.plot(Xn, np.average(Y_avg, axis=0), 'k',lw=3,label='avg fit')
plt.title(r"$\sigma$=%0.2f"%sigma)
plt.legend()
plt.xlabel(r"$X_n$")
plt.ylabel(r"$Y_{\mu}$")
#plt.show()

sigma = 0.5
plt.subplot(1,3,3)
Y_avg = []
for i in range(100):

    TH = np.random.choice(Theta, 100, replace=True)
    Yp = sum([TH[i] * Xn**i for i in range(len(TH))])
    TH,YP = map_estimate(Xn,Yp,sigma,alpha,d)
    Y_avg.append(YP)
    plt.plot(Xn,YP/10)

plt.plot(Xn, np.average(Y_avg, axis=0), 'k',lw=3,label='avg fit')
plt.title(r"$\sigma$=%0.2f"%sigma)
plt.legend()
plt.xlabel(r"$X_n$")
plt.ylabel(r"$Y_{\mu}$")
plt.show()

```
